```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python



df = pd.read_csv('/content/train.csv')


df.head()
```

```python

# Check missing values in every column

missing_values = df.isnull().sum()

# Display columns having missing values
missing_values = missing_values[missing_values > 0]

print("Missing Values:")
print(missing_values.sort_values(ascending=False))
```

```python

missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_table = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Missing Percentage': missing_percentage
})

missing_table = missing_table[missing_table['Missing Values'] > 0]

missing_table = missing_table.sort_values(
    by='Missing Percentage',
    ascending=False
)

missing_table
```

```python

plt.figure(figsize=(10, 5))

sns.histplot(df['SalePrice'], kde=True)

plt.title("Distribution of House Prices")
plt.xlabel("Sale Price")
plt.ylabel("Number of Houses")

plt.show()
```

```python

# Function to detect outliers using IQR

def detect_outliers(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = data[
        (data[column] < lower_limit) |
        (data[column] > upper_limit)
    ]

    print("Column:", column)
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
    print("Number of Outliers:", len(outliers))

# Check outliers
detect_outliers(df, 'SalePrice')
detect_outliers(df, 'GrLivArea')
```

```python

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.boxplot(y=df['SalePrice'])
plt.title("SalePrice Outliers")

plt.subplot(1, 2, 2)
sns.boxplot(y=df['GrLivArea'])
plt.title("Living Area Outliers")

plt.tight_layout()
plt.show()
```

```python

# Create a copy of the original dataset

df_clean = df.copy()

# Columns where missing means the feature is absent
absent_columns = [
    'PoolQC', 'Alley', 'Fence', 'FireplaceQu',
    'GarageType', 'GarageFinish', 'GarageQual',
    'GarageCond', 'BsmtQual', 'BsmtCond',
    'BsmtExposure', 'BsmtFinType1', 'BsmtFinType2',
    'MasVnrType'
]

for col in absent_columns:
    if col in df_clean.columns:
        df_clean[col] = df_clean[col].fillna('None')

# Fill LotFrontage using neighborhood median
df_clean['LotFrontage'] = df_clean.groupby(
    'Neighborhood'
)['LotFrontage'].transform(
    lambda x: x.fillna(x.median())
)

df_clean['LotFrontage'] = df_clean['LotFrontage'].fillna(
    df_clean['LotFrontage'].median()
)

# Fill remaining numerical missing values with median
numeric_cols = df_clean.select_dtypes(include=np.number).columns

for col in numeric_cols:
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())

# Fill remaining categorical missing values with mode
categorical_cols = df_clean.select_dtypes(include='object').columns

for col in categorical_cols:
    if df_clean[col].isnull().sum() > 0:
        df_clean[col] = df_clean[col].fillna(
            df_clean[col].mode()[0]
        )

print("Data Cleaning Completed!")
print("Remaining Missing Values:", df_clean.isnull().sum().sum())
```

```python

# Create final EDA dashboard

plt.figure(figsize=(16, 12))

# Graph 1: House Price Distribution
plt.subplot(2, 2, 1)

sns.histplot(df_clean['SalePrice'], kde=True)

plt.title("House Price Distribution")
plt.xlabel("Sale Price")


# Graph 2: Average Price by Neighborhood
plt.subplot(2, 2, 2)

neighborhood_price = df_clean.groupby(
    'Neighborhood'
)['SalePrice'].mean().sort_values()

neighborhood_price.tail(10).plot(kind='bar')

plt.title("Top 10 Expensive Neighborhoods")
plt.xlabel("Neighborhood")
plt.ylabel("Average Price")
plt.xticks(rotation=45)


# Graph 3: Living Area vs House Price
plt.subplot(2, 2, 3)

sns.scatterplot(
    data=df_clean,
    x='GrLivArea',
    y='SalePrice',
    hue='OverallQual'
)

plt.title("Living Area vs Sale Price")


# Graph 4: House Quality vs Sale Price
plt.subplot(2, 2, 4)

sns.boxplot(
    data=df_clean,
    x='OverallQual',
    y='SalePrice'
)

plt.title("House Quality vs Sale Price")
plt.xlabel("Overall Quality")
plt.xticks(rotation=45)


plt.tight_layout()

plt.savefig("House_Prices_EDA_Dashboard.png",
            dpi=300, bbox_inches='tight')

plt.show()
```
